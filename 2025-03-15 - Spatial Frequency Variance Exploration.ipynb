```python
SF=2.0**(arange(-4,4))
Ro=0
μe=3
μi=3
σe=1
σi=3

Ke=1/sqrt(2*pi*σe**2)*30
Ki=1/sqrt(2*pi*σi**2)*30

R=Ro+Ke*exp(-(SF-μe)**2/σe**2/2)-Ki*exp(-(SF-μi)**2/σi**2/2)

plot(SF,Ke*exp(-(SF-μe)**2/σe**2/2))
plot(SF,Ki*exp(-(SF-μi)**2/σi**2/2))
plot(SF,R,'-o')


idx_opt=argmax(R)
SFopt=SF[idx_opt]
M=16
idx=where((SF>(SFopt/M)) & (SF<=SFopt) )[0]

lSF=np.emath.logn(M,SF[idx])
lSFopt=np.emath.logn(M,SFopt)

LSFV=trapezoid(R[idx]*(lSF-lSFopt)**2,lSF)/trapezoid(R[idx],lSF)

title(LSFV)
```
